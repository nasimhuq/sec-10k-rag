```python
import sys

from mcp import Client, StdioServerParameters

from sec_10k_rag.common.config import ROOT_DIR
```

```python
SERVER = StdioServerParameters(
    command=sys.executable,
    args=["-m", "sec_10k_rag.step_08_mcp_server.run"],
    cwd=ROOT_DIR,
)
```

```python
async with Client(SERVER) as client:
    tools = await client.list_tools()
    tool = tools.tools[0]
    param_name = tool.input_schema["required"][0]
    param_type = tool.input_schema["properties"][param_name]["type"]
    result = await client.call_tool(tool.name, {param_name: "Nasim"})
```

```python
print(param_name, param_type)
```
